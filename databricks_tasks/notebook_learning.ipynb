**Task6:** 
Create a pyspark dataframe df1 reading the data from the above file using pyspark magic command %python

```python
%python

df1 = spark.read.csv(
    "/Volumes/workspace/default/mobile_metrics/mobile_os_usage.csv",
    header=True,
    inferSchema=True,
)

display(df1.limit(50))
```

**Task7:** 
Write the above dataframe df1 data into a databricks table called 'default.mobile_os_usage' using pyspark magic command %python

```python
%python
df1 = df1.withColumnRenamed(
    "Mobile Operating System",
    "Mobile_Operating_System"
)

df1 = df1.withColumnRenamed(
    "Percent of Usage",
    "Percent_of_Usage"
)
df1.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("default.mobile_os_usage")
display(spark.table("default.mobile_os_usage").limit(10))
```

**Task8:**
