### Widgets


Widgets help command

```python
dbutils.widgets.help()
```

#### Text widget

```python
dbutils.widgets.text("name", "vimal")
name = dbutils.widgets.get("name")
print(f"Hello {name}!!!")
```

#### Dropdown widgets

```python
dbutils.widgets.dropdown("city", "Chennai",["Chennai","Coimbatore","Madurai","Salem","Trichy"],"select city")
city = dbutils.widgets.get("city")
print(city)
```

#### Combobox widgets

```python
dbutils.widgets.combobox("country", "India",["India","Japan","Bolivia","Russia"],"select state")
country = dbutils.widgets.get("country")
print(country)
```

#### Multiselect Widget

```python
dbutils.widgets.multiselect("hobbies", "football",["cricket","football","swimming","gaming","reading","writing"],"select hobbies")
hobbies = dbutils.widgets.get("hobbies")
print(f"{hobbies} !!!")
```

Widgets Remove

```python
dbutils.widgets.remove("hobbies") # for one widget remove
dbutils.widgets.removeAll() # for all widgets remove
```
